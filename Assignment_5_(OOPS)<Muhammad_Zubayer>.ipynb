```python
class ShoppingCart:
  def __init__(self, customer_name, items = None):
    self.customer_name = customer_name
    if items is None:
      self.items = []
    else:
      self.items = items
  def add_item(self, price):
    if price > 0:
      self.items.append(price)
    else:
      print("invalid price.Please enter the Price")
  def remove_item(self, price):
    if price in self.items:
      self.items.remove(price)
    else:
      print("Your item isn't found.")
  def calculate_total(self):
     return sum(self.items)
  def apply_discount(self):
    total = self.calculate_total()
    if total>= 3000:
      total = total * .9
      return total
    else:
      print("You are not eligible for a discount.Please add more items to get discounts.")

  def display_cart(self):
    print(f"Customer Name: {self.customer_name}")
    print(f"Items in the cart: {self.items}")
    print(f"Total amount: {self.calculate_total()}")
    print(f"After Discount amount : {self.apply_discount()}")
    print(f" Number of items in the Cart : {len(self.items)}")
  def clear_cart(self):
    self.items.clear()
    print(f"Your cart has been cleared successfully.")

  # Part D: Combine Two Carts Using +
  def __add__(self, other):
        combined_items = self.items + other.items
        combined_customer_name = ( self.customer_name + " andd " + other.customer_name )
        return ShoppingCart( combined_customer_name, combined_items )


# 1.Creating Two Shopping Cart Objects
cart1 = ShoppingCart("Muhammad")
cart2 = ShoppingCart("Zubayer")

# 2.Adding and Removing Items from Both Carts
# Adding io cart-1
cart1.add_item(1200)
cart1.add_item(1500)
cart1.add_item(1300)
cart1.add_item(700)

# Adding in cart-2
cart2.add_item(1900)
cart2.add_item(5000)
cart2.add_item(1800)

# Removing from cart-1
cart1.remove_item(1300)

# Removing from cart-2
cart2.remove_item(1800)

# Displaying Individual Cart Details
cart1.display_cart()
print("\n")
cart2.display_cart()
print("\n")

# Combineding Two Carts
combined_cart = cart1 + cart2

# Displaying combined Cart Details
combined_cart.display_cart()
```
